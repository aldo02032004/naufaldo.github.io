````python
import pandas as pd

# Everything that used to be a copy-pasted config cell:
from great import MEDIA_MAP, plat_class, sent_class, plat_colors, sent_colors, validate_export
from great.viz import checks, prep
from great.viz.overview import weekly_overview
from great.viz.style import apply_style

apply_style()   # SciencePlots 'science' + 'no-latex', dpi 500
````

## 1. Load and validate

`validate_export()` fails loudly at load time if a column is missing or a
Sentiment/Media value is outside the canonical vocabulary — instead of that turning into a
silent `NaN` forty cells later.

Note the platform normalisation happens inside `prep.prepare_data()` via the library's
`MEDIA_MAP`, which covers `'Media Mainstream'` and `'Thread'`. The old notebook's local
copy did not, and would `KeyError` on `plat_colors` for exports containing them.

````python
import re
import requests
from io import BytesIO

SPREADSHEET_URL = "https://docs.google.com/spreadsheets/d/13jfQbm4DFMDSdtHN9aMHk5x83P4pahx4/edit?usp=sharing&ouid=114059382170725404105&rtpof=true&sd=true"

def download_google_sheet_as_xlsx(url):
    """Download isi Google Sheets sebagai .xlsx lewat link export publik.
    Butuh sharing setting: 'Anyone with the link' -> Viewer."""
    match = re.search(r"/d/([a-zA-Z0-9-_]+)", url)
    if not match:
        raise ValueError("Tidak bisa menemukan Sheet ID dari URL.")
    sheet_id = match.group(1)

    export_url = f"https://docs.google.com/spreadsheets/d/{sheet_id}/export?format=xlsx"
    response = requests.get(export_url)
    response.raise_for_status()
    return BytesIO(response.content)

excel_file = download_google_sheet_as_xlsx(SPREADSHEET_URL)
df = pd.read_excel(excel_file, header=1)
df = validate_export(df)

# Research object name, used in every plot title.
robj = 'DPR RI'

df.shape
````

## 2. Reshape

`prep` returns the three frames the plots consume. `prepare_data` also hands back the
period labels already formatted for the titles.

````python
df1, start_date, end_date = prep.prepare_data(df)

sent_data = prep.sentiment_data(df1)    # Date x Sentiment (wide)
plat_data = prep.platform_data(df1)     # (Date, Media) -> Count
ps_data   = prep.plat_sent_data(df1)    # Media, Sentiment, Count (long)

print(f'{start_date} - {end_date}')
````

## 3. Sanity checks

The same checks both notebooks ran inline, now as functions. They report rather than raise —
`validate_export()` above is the fail-fast version.

````python
checks.frame_info(df1, 'Main Dataframe')
checks.frame_info(sent_data, 'Sentiment Dataframe')
checks.frame_info(plat_data, 'Platform Dataframe')
checks.frame_info(ps_data, 'Platform-Sentiment Dataframe')
````

## 4. The overview figure

`weekly_overview` is the old `plot_sentiment_platform_overview`, with its three panels
factored into shared helpers that `monthly_overview` reuses.

`plat_class` / `sent_class` / `plat_colors` / `sent_colors` default to the library
vocabularies, so you only pass them to override. They are passed explicitly here to show
the full signature.

````python
sent_pct_df, platform_totals_df = weekly_overview(
    ps_data=ps_data,
    platform_data=plat_data,
    sentiment_data=sent_data,
    robj=robj,
    start_date=start_date,
    end_date=end_date,
    plat_class=plat_class,
    sent_class=sent_class,
    plat_colors=plat_colors,
    sent_colors=sent_colors,
)
````
